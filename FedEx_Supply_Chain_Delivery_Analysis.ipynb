## 7. Data Cleaning & Preprocessing

Data cleaning and preprocessing were performed to improve the quality, consistency, and usability of the dataset before conducting exploratory and statistical analysis.

The following preprocessing steps were carried out:

- **Creating a working copy:** A separate copy of the original dataset was created to preserve the raw data.
- **Date conversion:** Relevant shipment date columns were converted into datetime format to enable date-based calculations.
- **Numeric conversion:** Weight and freight cost columns were converted into numeric formats for quantitative analysis.
- **Duplicate checking:** Duplicate records were examined to avoid repeated observations affecting the analysis.
- **Missing-value handling:** Missing shipment mode values were categorized as "Unknown". Other missing values were retained and handled through analysis-specific filtering where required.
- **Feature engineering:** New variables, including Delivery Delay (Days), Delivery Status, and PO to Scheduled Delivery (Days), were created to support delivery performance analysis.
- **Data validation:** The cleaned dataset was checked for appropriate data types, remaining missing values, duplicate records, and consistency of derived variables.

These preprocessing steps established a consistent dataset for the subsequent exploratory data analysis, business-question investigation, and statistical testing.

```python
#4.1 Create a copy of the original dataset for cleaning
df_clean = df.copy()

print("Original dataset shape:", df.shape)
print("Working dataset shape:", df_clean.shape)
```

```python
# 4.2 Convert date columns into datetime format
date_columns = [
    'PQ First Sent to Client Date',
    'PO Sent to Vendor Date',
    'Scheduled Delivery Date',
    'Delivered to Client Date',
    'Delivery Recorded Date'
]

for col in date_columns:
    df_clean[col] = pd.to_datetime(df_clean[col], errors='coerce')

# Verify the conversion
df_clean[date_columns].dtypes
```

```python
#4.3 Convert weight and freight cost columns into numeric format
numeric_columns = [
    'Weight (Kilograms)',
    'Freight Cost (USD)'
]

for col in numeric_columns:
    df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')

# Check the updated data types
df_clean[numeric_columns].dtypes
```

```python
#4.4 Check for duplicate rows
duplicate_count = df_clean.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Remove duplicate records
df_clean = df_clean.drop_duplicates()

print("Dataset shape after duplicate removal:", df_clean.shape)
```

```python
#4.5 Recalculate missing values after data type conversion
missing_summary = pd.DataFrame({
    'Missing Values': df_clean.isnull().sum(),
    'Missing Percentage': (df_clean.isnull().sum() / len(df_clean)) * 100
})

# Display columns containing missing values
missing_summary = missing_summary[missing_summary['Missing Values'] > 0]

missing_summary.sort_values(
    by='Missing Percentage',
    ascending=False
)
```

```python
#5.3.1 Replace missing shipment modes with 'Unknown'
df_clean['Shipment Mode'] = df_clean['Shipment Mode'].fillna('Unknown')

# Verify the missing values have been handled
print("Missing values in Shipment Mode:",
      df_clean['Shipment Mode'].isnull().sum())

# Display shipment mode counts
df_clean['Shipment Mode'].value_counts()
```

```python
#5.3.2 Check missing values in numeric columns
numeric_columns = [
    'Weight (Kilograms)',
    'Freight Cost (USD)',
    'Line Item Insurance (USD)'
]

df_clean[numeric_columns].isnull().sum()
```

```python
#5.4.1 Calculate delivery delay in days
df_clean['Delivery Delay (Days)'] = (
    df_clean['Delivered to Client Date'] -
    df_clean['Scheduled Delivery Date']
).dt.days

# Display the result
df_clean[
    ['Scheduled Delivery Date',
     'Delivered to Client Date',
     'Delivery Delay (Days)']
].head(10)
```

```python
#5.4.2 Classify shipment delivery performance
df_clean['Delivery Status'] = np.select(
    [
        df_clean['Delivery Delay (Days)'] > 0,
        df_clean['Delivery Delay (Days)'] == 0,
        df_clean['Delivery Delay (Days)'] < 0
    ],
    [
        'Late',
        'On Time',
        'Early'
    ],
    default='Unknown'
)

# Check delivery status counts
df_clean['Delivery Status'].value_counts()
```

```python
#5.4.3 Calculate planned delivery duration in days
df_clean['PO to Scheduled Delivery (Days)'] = (
    df_clean['Scheduled Delivery Date'] -
    df_clean['PO Sent to Vendor Date']
).dt.days

# Display the calculated feature
df_clean[
    ['PO Sent to Vendor Date',
     'Scheduled Delivery Date',
     'PO to Scheduled Delivery (Days)']
].head(10)
```

## 9. Statistical Analysis

```python
#6.7 — Statistical test: Is Shipment Mode related to Delivery Status?


#6.7.1 Create contingency table for Shipment Mode and Delivery Status

contingency_table = pd.crosstab(
    df_clean['Shipment Mode'],
    df_clean['Delivery Status']
)

contingency_table
```

```python
from scipy.stats import chi2_contingency

#6.7.2 Perform Chi-square test
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("Chi-square Statistic:", round(chi2, 4))
print("P-value:", p_value)
print("Degrees of Freedom:", dof)

print("\nExpected Frequencies:")
print(pd.DataFrame(
    expected,
    index=contingency_table.index,
    columns=contingency_table.columns
).round(2))
```

```python
#6.7.3 Calculate Cramer's V effect size

n = contingency_table.to_numpy().sum()
min_dimension = min(contingency_table.shape[0] - 1,
                   contingency_table.shape[1] - 1)

cramers_v = np.sqrt(chi2 / (n * min_dimension))

print("Cramer's V:", round(cramers_v, 4))
```

```python
#6.8 — Statistical test: Country vs Delivery Status

#6.8.1 Select countries having at least 50 shipments
eligible_countries = (
    df_clean['Country']
    .value_counts()
)

eligible_countries = eligible_countries[
    eligible_countries >= 50
].index

# Filter dataset for eligible countries
country_test_data = df_clean[
    df_clean['Country'].isin(eligible_countries)
]

# Create contingency table
country_contingency = pd.crosstab(
    country_test_data['Country'],
    country_test_data['Delivery Status']
)

country_contingency
```

```python
from scipy.stats import chi2_contingency

#6.8.2 Perform Chi-square test
chi2_country, p_country, dof_country, expected_country = (
    chi2_contingency(country_contingency)
)

print("Chi-square Statistic:", round(chi2_country, 4))
print("P-value:", p_country)
print("Degrees of Freedom:", dof_country)

print("\nMinimum Expected Frequency:",
      expected_country.min().round(2))

print("\nExpected Frequencies:")
pd.DataFrame(
    expected_country,
    index=country_contingency.index,
    columns=country_contingency.columns
).round(2)
```

```python
#6.8.3 Calculate Cramer's V for Country and Delivery Status

n_country = country_contingency.to_numpy().sum()

min_dim_country = min(
    country_contingency.shape[0] - 1,
    country_contingency.shape[1] - 1
)

cramers_v_country = np.sqrt(
    chi2_country / (n_country * min_dim_country)
)

print("Cramer's V:", round(cramers_v_country, 4))
```

```python
from scipy.stats import chi2_contingency

#6.9.4 Perform Chi-square test
chi2_vendor, p_vendor, dof_vendor, expected_vendor = chi2_contingency(
    vendor_contingency
)

print("Chi-square Statistic:", round(chi2_vendor, 4))
print("P-value:", p_vendor)
print("Degrees of Freedom:", dof_vendor)

# Check expected frequency assumptions
print("\nMinimum Expected Frequency:",
      round(expected_vendor.min(), 2))

print("Expected cells below 5:",
      (expected_vendor < 5).sum())

print("Percentage of expected cells below 5:",
      round((expected_vendor < 5).mean() * 100, 2), "%")
```

```python
#6.9.5 Calculate Cramer's V

n_vendor = vendor_contingency.to_numpy().sum()

min_dim_vendor = min(
    vendor_contingency.shape[0] - 1,
    vendor_contingency.shape[1] - 1
)

cramers_v_vendor = np.sqrt(
    chi2_vendor / (n_vendor * min_dim_vendor)
)

print("Cramer's V:", round(cramers_v_vendor, 4))
```

## 11. Key Findings & Recommendations

## Final Findings
The analysis of the FedEx Supply Chain delivery dataset, consisting of 10,324 shipment records, revealed several important patterns related to delivery performance, shipment characteristics, and supplier reliability.

**1. Overall Delivery Performance:** Approximately 61.26% of shipments were delivered on time, while 27.26% arrived early and 11.49% were delivered late. Although most shipments met or exceeded their scheduled delivery dates, the late-delivery rate indicates opportunities for improving logistics performance.

**2. Shipment Mode:** Delivery performance varied across shipment modes. Ocean shipments recorded the highest late-delivery rate of 17.52%, followed by Truck at 16.08%, Air Charter at 11.54%, and Air at 9.60%. This suggests that delivery outcomes are associated with the shipment mode selected.

**3. Country-Level Performance:** Delivery delays varied significantly across destination countries. Burundi recorded the highest late-delivery rate of 38.78% among countries with at least 50 shipments, followed by Congo, DRC at 24.92% and Mozambique at 18.38%. These destinations may require additional monitoring and logistics planning.

**4. Team Management:** PMO-US managed 10,265 out of 10,324 shipments and recorded an on-time delivery rate of 61.04%. Other management teams had considerably smaller shipment volumes, making direct comparisons less reliable. The results primarily reflect PMO-US performance.

**5. Planned Lead Time:** Late shipments had a higher median planned lead time of 107 days compared with 92 days for on-time shipments and 90 days for early shipments. However, the substantial variation in lead time indicates that planned delivery time alone may not explain delivery outcomes.

**6. INCO Terms:** Delivery performance varied across INCO terms. EXW recorded a late-delivery rate of 4.97%, DDP 8.04%, and FCA 1.01%. The N/A – From RDC category had a late-delivery rate of 17.15%, indicating a distinct delivery-performance pattern. Categories with small shipment counts should be interpreted cautiously.

**7. Shipment Weight and Insurance Cost:** Shipment weight showed a positive association with line item insurance cost. Pearson correlation was 0.6172, while Spearman correlation was 0.7786, suggesting that heavier shipments generally tend to have higher insurance costs. Nevertheless, variability in insurance cost indicates that other factors may also contribute.

**1. Improve Monitoring of Delayed Shipments:** Implement a shipment tracking and early-warning system to identify shipments at risk of missing scheduled delivery dates. Priority attention should be given to shipments with a history of delays.

**2. Optimize Shipment Mode Selection:** Review the performance of Ocean and Truck shipments, which recorded higher late-delivery rates. Where feasible, evaluate alternative shipment modes based on urgency, cost, destination, and delivery reliability.

**3. Focus on High-Risk Destinations:** Countries with elevated late-delivery rates, such as Burundi, Congo, DRC, and Mozambique, should receive additional logistical attention. High-performing vendors can be used as benchmarks, while vendors with frequent delays should be reviewed to identify improvement opportunities.
